# View an object in a notebook (Jupyter, Colab, RStudio)

<a href="https://colab.research.google.com/github/theislab/cellucid-python/blob/main/docs/user_guide/r_package/f_notebooks_tutorials/07_view_in_a_notebook.ipynb"><img src="https://colab.research.google.com/assets/colab-badge.svg" alt="Open In Colab"></a>

One line shows a Seurat, SingleCellExperiment or SummarizedExperiment
object, or any file Cellucid reads, in the Cellucid viewer: in the
output of a Jupyter notebook cell (JupyterLab, Notebook 7, VS Code,
Google Colab's R runtime), in the RStudio or Positron Viewer pane, or
in a browser tab from any other R console. The same code, everywhere.

**Time:** 5 minutes. **You need:** the `cellucid` package; in Jupyter
also `IRkernel` and `IRdisplay`, and `anywidget` where the notebook
server runs (`pip install anywidget`; VS Code and Colab bring their
own widget support). The browser loads the app from
https://www.cellucid.com; your data does not go there.

The same tutorial in Python: [Show a dataset in any notebook](../../python_package/f_notebooks_tutorials/41_show_a_dataset_in_any_notebook.ipynb).

```r
# A fresh R library (Google Colab's R runtime, a new machine) installs the packages once.
for (package in c("remotes", "SeuratObject")) {
  if (!requireNamespace(package, quietly = TRUE)) install.packages(package)
}
if (!requireNamespace("cellucid", quietly = TRUE)) {
  if ("cellucid" %in% rownames(available.packages())) install.packages("cellucid") else remotes::install_github("theislab/cellucid-r")
}
```

## The data

SeuratObject's own example, `pbmc_small`: 80 blood cells, 230 genes, a
t-SNE and the clusters `RNA_snn_res.1`. Use your own Seurat or
SingleCellExperiment object in its place.

```r
library(cellucid)
library(SeuratObject)
data("pbmc_small")
seu <- pbmc_small
seu
```

## The one line

`dataset_id` and `dataset_name` are optional: without them the id is
made from the object's name, and a message says so.

```r
viewer <- cellucid_view(seu, dataset_id = "pbmc-small", dataset_name = "PBMC small")
```

Where the viewer appears:

| Where you run R | The viewer |
|---|---|
| a Jupyter notebook (JupyterLab, Notebook 7, VS Code, Google Colab; IRkernel) | in the cell's output, once the cell finishes (a notebook widget over the kernel's own connection: no port, no server) |
| RStudio, Positron | the Viewer pane |
| any other R console | a browser tab |

In a notebook, R hears the viewer **between cells**: run what reads the
viewer's work (`cellucid_selection()`, `cellucid_pull()`) in a later cell
than the one that showed it.

## Drive the viewer from R

Commands go out at once (one sent before the viewer has loaded waits
and is delivered when it connects). Cells are named by cell name, by
1-based position, or as `TRUE`/`FALSE` per cell.

```r
cellucid_color_by(viewer, "RNA_snn_res.1")
cellucid_highlight(viewer, WhichCells(seu, idents = "1"), color = "#e41a1c", label = "Cluster 1 (from R)")
```

Now lasso some cells in the viewer: **Highlighting › Lasso**, drag
around a group, **Confirm**. Then run the next cell.

```r
selected <- cellucid_selection(viewer)  # NULL until you confirm a selection
if (!is.null(selected)) {
  print(length(selected$cell_names))
  print(table(seu$groups[selected$cell_names]))
}
```

Or react to every selection as it is made:

```r
cellucid_on(viewer, "selection", function(event) {
  message(length(event$cells), " cells selected with the ", event$source)
})
```

## If nothing shows

In Jupyter, `anywidget` must be installed where the notebook *server*
runs (JupyterLab, Notebook 7) and the browser must reach
www.cellucid.com (or a copy you host: `web_source_url =`). Only
JupyterLab is tested by the project; Notebook 7, VS Code and Colab
show the same widget protocol. In Colab's R runtime, `cellucid_view()`
names Colab's widget manager itself (the project has checked the
messages, not Colab itself): if no viewer appears there, write the
object with `cellucid_prepare()` and open the file at
www.cellucid.com. More: [View in Jupyter notebooks](../d_viewing_loading/08_view_in_jupyter_notebooks.md).

## Stop

```r
cellucid_stop(viewer)
```
